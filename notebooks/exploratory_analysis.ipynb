# Crop Recommendation Dataset: Exploratory Analysis

This notebook analyzes the supplied real CSV only. Place it at `../dataset/Crop_recommendation.csv`. No synthetic data or model results are generated.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

csv_path = Path('../dataset/Crop_recommendation.csv')
if not csv_path.exists():
    raise FileNotFoundError('Download the Kaggle CSV and place it in dataset/.')
df = pd.read_csv(csv_path)
df.head()
```

```python
required = ['N','P','K','temperature','humidity','ph','rainfall','label']
print('Shape:', df.shape)
print('Missing cells:', int(df.isna().sum().sum()))
print('Duplicate rows:', int(df.duplicated().sum()))
print('Types:\n', df.dtypes)
df['label'].value_counts().sort_index()
```

```python
df['label'].value_counts().sort_index().plot(kind='bar', figsize=(12,4), color='seagreen')
plt.title('Crop class distribution')
plt.xlabel('Crop'); plt.ylabel('Rows'); plt.tight_layout()
```
